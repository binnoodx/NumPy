```python
import numpy as np
```

```python
#Reshaping in numpy
arr1 = np.arange(6).reshape(2,3)
```

```python
arr2 = np.arange(6,12).reshape(3,2)
```

```python
#Changing 2-d array to 1-d
arr1.ravel()
```

```python
#Transposing the array
arr2.transpose()
```

```python
#splitting

arr3 = np.arange(9).reshape(3,3)
```

```python
arr3
```

```python
np.hsplit(arr3 , 3) #Horizontal Splitting
```

```python
np.vsplit(arr3 , 3) #Vertical Splitting
```

```python
#stacking
arr1 = np.array([[1,2,3] , [4,5,6]])
```

```python
arr2 = np.array([[6,7,8] , [8,9,10]])
```

```python
np.hstack((arr1 , arr2))
```

```python
np.vstack((arr1 , arr2))
```
