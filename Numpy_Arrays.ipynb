```python
import numpy as np
```

```python
arr1 = np.array([1,2,3,4,5])
#1-d array
arr1
```

```python
arr2 = np.array([[1,2,3] , [4,5,6]])
#2-d array
arr2
```

```python
arr3 = np.zeros((2,3))
#2X3 order filled with zeros
arr3
```

```python
arr4 = np.ones((3,3))
#3x3 order filled with ones
arr4
```

```python
arr5 = np.identity(5)
#identity matrix of order 5X5
arr5
```

```python
arr6 = np.arange(10)
#Array from 0-9
arr6
```

```python
arr7 = np.linspace(2,8,4)
#generates array of 4 numbers starts from 2 to 8 having even space between them 
arr7
```

```python
arr8 = arr7.copy()
#Copy the Array
arr8
```
