```python
import numpy as np
```

```python
arr1 = np.random.randint(low=1 , high=100 , size=20).reshape(4,5)
```

```python
arr1
```

```python
arr1[ :, [0,3]] # Fancy Indexing-----> Getting first and fourth columns
```

```python
arr1[[0,2]]  # Fancy Indexing-----> Getting first and third rows
```

```python

#------Indexing with boolean arrays

arr1[arr1>50] #gives array of those num greater than 50
```

```python
arr1[(arr1>50) & (arr1%2==0)]
```

```python
arr1[arr1%2!=0] = 69 #make all odd numbers of array to 69
```
