```python
import numpy as np
```

```python


marks = np.array([
    [78, 85, 90],   # Student 1
    [45, 50, 40],   # Student 2
    [88, 92, 95],   # Student 3
    [60, 70, 65],   # Student 4
    [30, 25, 35]    # Student 5
])

#Pass Marks = 150
```

```python
total = marks.sum(axis=1 , )
print("The Total Marks of each students is" , total)
```

```python
high_scorer = np.argmax(total)
print("The High Scorer is Student", high_scorer+1 , "and his marks is", total[high_scorer])
```

```python
low_scorer = np.argmin(total)
print("The Low Scorer is Student", low_scorer+1 ,"and his marks is", total[low_scorer])
```

```python
passedStudents = np.where(total>=150)
print("The Passed Students are ")
for i in np.nditer(passedStudents):
    print( i+1)
```

```python
failedStudents = np.where(total<=150)
print("The Failed Students are ")
for i in np.nditer(failedStudents):
    print( i+1)
```

```python
passing_percent = (np.size(passedStudents))/5 * 100
print(passing_percent, "% Student had passed.")
```

```python
failing_percent = (np.size(failedStudents))/5 * 100
print(failing_percent, "% Student had failed.")
```
