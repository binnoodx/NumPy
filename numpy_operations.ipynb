```python
import numpy as np
```

```python
#Arithmetic Operations
arr1 = np.array([1,2,3,4,5])
```

```python
arr2 = np.array([5,6,7,8,9])
```

```python
arr1 + arr2
```

```python
arr1-arr2
```

```python
arr1 * arr2 #vector multipication
```

```python
arr1 * 2 #Scaler Multipication
```

```python
#Dot Product between different order matrices
arr3 = np.arange(6).reshape(2,3)
```

```python
arr4 = np.arange(6,12).reshape(3,2)
```

```python
arr3.dot(arr4)
```

```python
arr1 > 2 #Conditional operations
```

```python
#Statistics Operations
arr1.mean() #Mean
```

```python
arr1.std() #Standard Deviation
```

```python
np.median(arr1)
```

```python
#Math Operations
np.sin(arr1)
```

```python
np.exp(arr1)
```

```python
np.sqrt(arr1) #-------> Instance Function
```

```python
arr1.max()  #------> Static Function
```
