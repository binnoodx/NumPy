```python
#list in python
list1 = range(100) #more memory , slower , less convenient
#array in numpy
arr1 = np.arange(100) #less memory , faster , more convinent

```

```python
#Time Calculation
import time
```

```python
a = range(10000000)
b = range(10000000 , 20000000)
start_time = time.time()
c = [a+b for a,b in zip(a,b)] #less conv
print(time.time() - start_time) #0.89 sec
```

```python
x = np.arange(10000000)
y = np.arange(10000000 , 20000000)
start_time = time.time()
c = x + y
print(time.time() - start_time) #0.17 sec
```
